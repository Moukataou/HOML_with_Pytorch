```python
from sklearn.datasets import fetch_covtype

# Dataset path for global access
DATA_HOME = "../datasets"

# Loading dataset
covtype = fetch_covtype(data_home=DATA_HOME)

X, y = covtype.data, covtype.target
```

```python
from sklearn.metrics import accuracy_score
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

# Shuffling and splitting dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

# Creating Model
MLP_clf = make_pipeline(StandardScaler(), MLPClassifier(verbose=True, early_stopping=True))
```

```python
# Tuning hyperparameters 1
param_grid = {'mlpclassifier__hidden_layer_sizes': ((100,), (100, 100), (100, 100, 100))}
grid_search = GridSearchCV(MLP_clf, param_grid, cv=3, scoring='accuracy', n_jobs=-1)
grid_search.fit(X_train, y_train)
```

```python
# Results 1
print("Best parameters:", grid_search.best_params_, "with accuracy:", grid_search.best_score_)
```

```python
# Tuning hyperparameters 2
param_grid = {'mlpclassifier__hidden_layer_sizes': ((100, 100, 100), (300, 300, 300), (500, 500, 500))}
grid_search = GridSearchCV(MLP_clf, param_grid, cv=3, scoring='accuracy', n_jobs=-1)
grid_search.fit(X_train, y_train)
```

```python
# Results 2
print("Best parameters:", grid_search.best_params_, "with accuracy:", grid_search.best_score_)
```

```python
# Evaluating the best model on the test set
best_model = grid_search.best_estimator_
y_pred = best_model.predict(X_test)
print("Accuracy sur le test set :", accuracy_score(y_test, y_pred))
```
